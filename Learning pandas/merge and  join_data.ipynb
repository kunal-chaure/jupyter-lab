```python
import pandas as pd

# Customer Data
df_customers = pd.DataFrame({
    "customer_id": [1, 2, 3, 4],
    "name": ["Adam", "Bob", "Charlie", "Dave"]
})

# Orders Data
df_orders = pd.DataFrame({
    "order_id": [101, 102, 103, 104],
    "customer_id": [2, 1, 4, 5],
    "amount": [250, 120, 300, 180]
})
```

```python
pd.merge(df_customers,df_orders )
pd.merge(df_customers,df_orders, on = "customer_id")   #this is called inner join
pd.merge(df_customers,df_orders, on = "customer_id", how="left") # left join
pd.merge(df_customers,df_orders, on = "customer_id",  how ="right")
pd.merge(df_customers,df_orders, on = "customer_id",  how ="outer")
```
