```python
import pandas as pd

df = pd.read_csv("raw_data.csv")
```

```python
df.isnull()  #we can also use isna() same meaning 

df.isnull().sum()     #returns the sum of all NaN

df.dropna()

df.dropna(axis = 1)
# df.fillna(0.0)
```

```python
# cleaning the data 
# age_mean = df["age"].mean()

clean_data = df.copy()
age_mean = clean_data["age"].mean()
clean_data["age"] = clean_data["age"].fillna(age_mean)

clean_data
```
