```python
import pandas as pd
```

```python
df = pd.DataFrame({
    "Country":["USA","USA","INDIA","INDIA"],
    "Year":[2020,2021,2020,2021],
    "Sales":[100,120,90,110],
    "Profits":[20,25,18,22]
})
df.melt(
    id_vars= ["Country","Year"],
    value_vars=["Sales","Profits"],
    var_name="metrics",
    value_name="value"

)
```
