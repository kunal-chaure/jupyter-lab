```python
import numpy as np
import time
```

```python
# execution performence
size = 1_00_000

py_list=list(range(size))
start = time.time()
square_list= [ x**2 for x in py_list]
end= time.time()
print(f"python list time ={end-start} second")

# numpy  array
np_arr = np.array(py_list)

start = time.time()
sq_arr= np_arr**2
end= time.time()
print(f"numpy array list time ={end-start} second")
```

```python
# memory  occupied
import sys
print(f"the memory occupied by list or python list size {sys.getsizeof(py_list)*len(py_list)} bytes")
print(f"numpy array size = {np_arr.nbytes} bytes")
```
