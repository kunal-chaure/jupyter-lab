```python
import numpy as np
```

```python
# creating array using in build method
arr1 = np.zeros((2,3), dtype="int64")
print(arr1,arr1.shape) 
# this is called prefill

arr1 = np.ones((7,30), dtype="int64")
print(arr1,arr1.shape)


arr1 = np.full((3,4),100)
print(arr1,arr1.shape)#this is prefill with value

arr1 = np.eye(3)
print(arr1,arr1.shape)#this is  called identical metricx



arr = np.arange(1,20,1)
print(arr,arr.shape)#this is  

arr = np.linspace(1,200,5)
print(arr,arr.shape)#this is
```

```python
# now learning about array properties
arr = np.array([1,2,4,5,6,7,8,9,0])
print(arr.shape)
print (arr.size)
print (arr.dtype)
print(arr.ndim)

float_arr = arr.astype(np.float64)
print(float_arr , float_arr.dtype)
```

```python
# reshaping the array
import numpy as np
arr = np.array([[1,3,4],[5,6,7]])
# print(arr)
reshape = arr.reshape((3,2))
print(reshape)

flattened = arr.flatten()
print(flattened)
```

```python
import numpy  as np
# going to learn about indexing on array
arr = np.array([1,3,4,5,6,7,8])
print(arr[0])
print(arr[1])
print(arr[5])
print(arr[3])
print(arr[6])

# now on 2d array
arr = np.array([[1,3,4],[5,6,7]])


print(f"the no is",arr[1][1])
print(f"the no is",arr[1][0])
print(f"the no is",arr[0][0])

# now learn about fancy indexing
arr = np.array([1,3,4,5,6,7,8])

idx= [0,1,3]
print(arr[idx])

# boolean indexing
arr = np.array([1,3,4,5,6,7,8])
print(arr [arr>2])
print(arr [arr%2==0])
```

```python
# now learning about slice in array
import numpy as np

arr = np.array([1,3,4,5,6,7,8])

print(arr[:])
print(arr[1:])
print(arr[:3])
print(arr[3:])
print(arr[::2])
```
