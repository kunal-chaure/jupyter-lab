```python
# mathematical funtion 

# now learning about aggregate funtion

    import numpy as np
    arr = np.array([1,2,3,4,5])
    print (np.sum(arr))
    print (np.prod(arr))
    print (np.min(arr))
    print (np.max(arr))
    print (np.argmin(arr))
    print (np.argmax(arr))
    
    print (np.mean(arr))
    print (np.std(arr))
    print (np.median(arr))
    print (np.var(arr)) # this is about varience
```

```python
import numpy as np
# this are the power functions
arr = np.array([1,2,3,4,5])
print (np.square(arr))
print (np.sqrt(arr))
print (np.pow(arr,3))
print ( np.log(arr))
```
