```python
import json
```

```python
# load the data
def load_data(filename):
    with open(filename,"r") as f:
        data = json.load(f)
        return data
```

```python
data = load_data("store_data.json")
print(data)
```

```python
#clean the data
def clean_data(data):
    text_to_no = {"one":1,"two":2,"three":3,"four":4,"five":5}
    cleaned_data = []
    unique_users = set()
    
    for user in data:
        # print(user)
        raw_rating= user["rating"].strip().lower()
        # print(raw_rating)
        if (raw_rating in text_to_no):
            raw_rating = text_to_no[raw_rating]
        print(raw_rating)
        user["rating"] = raw_rating
       
     
        # handle missing values
        raw_age =  user.get("age")
        if (raw_age == None):
            user["age"]=None
        # deduplication
        if(user["name"].strip() in unique_users):
            continue
        unique_users.add(user["name"])
        cleaned_data.append(user) 
    return cleaned_data
```

```python
 data = clean_data(data)
```

```python
get_insights(data)
```

```python
#Build recommendation feature
def get_recommendation(data):
    recommendations = []

    for user in data:
        curren_recom ={}
        curren_recom["name"] = user["name"]

        if(float (user["rating"])>= 4 ):
            curren_recom["brand"]="Apple"
        else:
            curren_recom["brand"] = "samsung"
        recommendations.append(curren_recom)
    return recommendations
```

```python
get_recommendation(data)
```
